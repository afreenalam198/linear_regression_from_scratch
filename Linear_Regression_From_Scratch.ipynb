# **Linear Regression From Scratch (Single Feature)**

This section demonstrates a single-feature Linear Regression model solved via iterative Optimization (Gradient Descent).

### Mathematical Foundations:
1. **Hypothesis**: $y_{\text{pred}} = wX + b$
2. **Loss Function (Mean Squared Error)**:
   $$\text{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - y_{\text{pred}, i})^2$$
3. **Gradient Calculations (Partial Derivatives with respect to $w$ and $b$ )**:
   $$\frac{\partial L}{\partial w} = -\frac{2}{N} \sum_{i=1}^{N} X_i(y_i - y_{\text{pred}, i})$$
   $$\frac{\partial L}{\partial b} = -\frac{2}{N} \sum_{i=1}^{N} (y_i - y_{\text{pred}, i})$$
4. **Parameter Updates**:
   $$w \leftarrow w - \alpha \frac{\partial L}{\partial w}$$
   $$b \leftarrow b - \alpha \frac{\partial L}{\partial b}$$
   *where $\alpha$ is the learning rate.*

```python
import numpy as np

# Dataset (3 houses)
X = np.array([1, 2, 3]) # Bedrooms
y_true = np.array([60, 115, 150]) # Actual Prices ($k)

# Model's current parameters
w = 50
b = 10
learning_rate = 0.01

# Forward Pass
def predict(X, w, b):
    # Implement the linear regression hypothesis mathematically
    y_pred = w * X + b
    return y_pred

# Loss
def mse_loss(y_true, y_pred):
    # Implement the Mean Squared Error formula
    mse = np.mean((y_true - y_pred)**2)
    return mse

# Gradient Descent
def update_params(X, y_true, y_pred, w, b, learning_rate):
  # Calculate Gradient of Weight
  dw = -2 * np.mean(X * (y_true - y_pred))

  # Calculate Gradient of Bias
  db = -2 * np.mean(y_true - y_pred)

  w_new = w - (learning_rate * dw)
  b_new = b - (learning_rate * db)

  return w_new, b_new

# Model prediction and loss:
y_pred = predict(X, w, b)
loss = mse_loss(y_true, y_pred)
print(f"Predictions: {y_pred}")
print(f"MSE Loss: {loss}")

# Updated Weight and Bias
w_new, b_new = update_params(X, y_true, y_pred, w, b, learning_rate)
print(f"New Weight: {w_new}, New Bias: {b_new}")
```

# **Complete Linear Regression Iterative Training Loop**

Here we encapsulate our Gradient Descent algorithm within an object-oriented Python class mimicking the Scikit-Learn API structure (`fit` and `predict` paradigm).

This implementation updates the parameters over many sequential iterations (epochs) to minimize the Mean Squared Error.

```python
class LinearRegression:
  def __init__(self, learning_rate: float = 0.01, epochs: int = 1000) -> None:
    self.lr = learning_rate
    self.epochs = epochs
    # Model's state
    self.w = 0.0
    self.b = 0.0

  def predict(self, X: np.ndarray) -> np.ndarray:
      # Implement the forward pass (linear regression hypothesis mathematically)
      y_pred = self.w * X + self.b
      return y_pred

  def mse_loss(self, y_true: np.ndarray, y_pred: np.ndarray) -> float:
      # Implement the Mean Squared Error
      mse = np.mean((y_true - y_pred)**2)
      return mse

  def fit(self, X: np.ndarray, y_true: np.ndarray) -> None:
    for epoch in range(self.epochs):
      # Get predictions
      y_pred = self.predict(X)

      # Calculate Gradient of Weight
      dw = -2 * np.mean(X * (y_true - y_pred))

      # Calculate Gradient of Bias
      db = -2 * np.mean(y_true - y_pred)

      # Update self.w and self.b
      self.w = self.w - (self.lr * dw)
      self.b = self.b - (self.lr * db)

      # Logging progress
      if epoch % 200 == 0:
        loss = self.mse_loss(y_true, y_pred)
        print(f"Epoch {epoch:4d} | Loss: {loss:.4f} | w: {self.w:.4f}, b: {self.b:.4f}")

if __name__ == '__main__':
  # Dataset: Experience (Years) vs. Salary ($10k units)
  X = np.array([1.1, 2.0, 3.2, 5.1, 8.2, 10.5])
  y_true = np.array([3.9, 4.3, 5.5, 7.1, 10.5, 12.2])

  # Instantiate the model
  model = LinearRegression(learning_rate=0.01, epochs=1000)

  # Train the model
  print("Starting training...")
  model.fit(X, y_true)
  print("--- Training Complete ---")

  # Inference (Make a prediction)
  new_candidate_xp = np.array([5.0])
  predicted_salary = model.predict(new_candidate_xp)
  print(f"\nPredicted Salary for 5 years XP: ${predicted_salary[0] * 10000:,.2f}")
```

# **Linear Regression Using the Closed-Form Normal Equation**

Instead of iteratively searching for the global minimum via Gradient Descent, we can compute the analytical solution directly using the **Normal Equation**.

### Mathematical Derivation:
$$\theta = (X^T X)^{-1} X^T y$$

### The Bias Trick:
To compute the bias ($b$) and weights ($w$) simultaneously, we append an additional vector of ones ($1$) as the first column of our feature matrix $X$ (representing the constant intercept factor):
$$X = \begin{bmatrix} 1 & x_{1} \\ 1 & x_{2} \\ \vdots & \vdots \\ 1 & x_{n} \end{bmatrix}$$

*Note: Calculating the matrix inverse $(X^T X)^{-1}$ has a complexity of  ${O}(D^3)$, making this approach highly efficient for small-to-medium feature spaces, but computationally expensive as the number of dimensions $D$ scales.*

```python
import numpy as np

# Dataset: Years of Experience vs Salary ($10k units)
X_raw = np.array([[1.1], [2.0], [3.2], [5.1], [8.2], [10.5]])
y = np.array([3.9, 4.3, 5.5, 7.1, 10.5, 12.2])

# The Bias Trick: Add a column of 1s to the left of X
X = np.c_[np.ones(X_raw.shape[0]), X_raw]

def normal_equation(X, y):
    # Implement W = (X^T * X)^-1 * X^T * y
    # 1. Use X.T for transpose
    # 2. Use np.dot(A, B) or the @ operator (A @ B) for matrix multiplication
    # 3. Use np.linalg.inv(matrix) to calculate the inverse

    W = np.linalg.inv(X.T @ X) @ X.T @ y
    return W

# Calculate weights instantly
W_optimal = normal_equation(X, y)

# The first number in W is the bias (b), the second is the weight (w)
bias = W_optimal[0]
weight = W_optimal[1]

print(f"Analytical Bias: {bias:.4f}")
print(f"Analytical Weight: {weight:.4f}")
```

# **Evaluating Scratch vs. Scikit-Learn on the Diabetes Dataset**

This benchmark validates our custom vectorised Gradient Descent implementation against Scikit-Learn's optimized solver using real-world medical data.

### Engineering Best Practice - Feature Scaling:
Because Gradient Descent update step magnitudes depend heavily on feature scales, we apply **Z-Score Standardization** (Zero-Mean, Unit-Variance) before training to ensure a spherical loss landscape and stable, fast convergence:
$$X_{\text{scaled}} = \frac{X - \mu}{\sigma}$$

We verify our implementation by comparing final converged weights, bias, and test Mean Squared Error (MSE) metrics.

```python
# Import Libraries
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression as SklearnLR
from sklearn.metrics import mean_squared_error

# From Scratch Model
class ScratchLinearRegression:
  def __init__(self, learning_rate: float = 0.01, epochs: int = 1000) -> None:
    self.lr = learning_rate
    self.epochs = epochs
    # Model's state
    self.w = None
    self.b = 0.0

  def predict(self, X: np.ndarray) -> np.ndarray:
      # Implement the forward pass (linear regression hypothesis mathematically)
      y_pred = np.dot(X, self.w) + self.b
      return y_pred

  def mse_loss(self, y_true: np.ndarray, y_pred: np.ndarray) -> float:
      # Implement the Mean Squared Error
      mse = np.mean((y_true - y_pred)**2)
      return mse

  def fit(self, X: np.ndarray, y_true: np.ndarray) -> None:
    n_samples, n_features = X.shape

    # Intialize Weight matrix with a weight of 0.0 for each feature
    self.w = np.zeros(n_features)

    for epoch in range(self.epochs):
      # Get predictions
      y_pred = self.predict(X)

      # Calculate gradients of Weight matrix
      dw = (-2 / n_samples) * (np.dot(X.T, (y_true - y_pred)))

      # Calculate gradient of Bias
      db = -2 * np.mean(y_true - y_pred)

      # Update self.w and self.b
      self.w = self.w - (self.lr * dw)
      self.b = self.b - (self.lr * db)

if __name__ == '__main__':
  # Load and preprocess medical dataset
  print("Fetching Diabetes Dataset...")
  data = load_diabetes(as_frame=True)
  df = data.frame

  # Using 4 features from the dataset
  features = ['age', 'sex', 'bmi', 'bp']
  X_raw = df[features].values
  y_raw = df['target'].values

  # Standardize features using Z-Score
  X_mean = np.mean(X_raw, axis=0)
  X_std = np.std(X_raw, axis=0)
  X_scaled = (X_raw - X_mean) / X_std

  y_mean = np.mean(y_raw)
  y_std = np.std(y_raw)
  y_scaled = (y_raw - y_mean) / y_std

  # Train ScratchLinearRegression Model
  print("\nTraining Scratch Model...")
  scratch_model = ScratchLinearRegression(learning_rate=0.1, epochs=1000)
  scratch_model.fit(X_scaled, y_scaled)
  scratch_preds = scratch_model.predict(X_scaled)
  scratch_mse = scratch_model.mse_loss(y_scaled, scratch_preds)

  # Train Sckiti-Learn Model
  print("Training Scikit-Learn Model...")
  sklearn_model = SklearnLR()
  sklearn_model.fit(X_scaled, y_scaled)
  sklearn_preds = sklearn_model.predict(X_scaled)
  sklearn_mse = mean_squared_error(y_scaled, sklearn_preds)

  print("\n--- RESULTS COMPARISON ---")
  print(f"{'Metric':<15} | {'Scratch Model':<25} | {'Scikit-Learn':<25}")
  print("-" * 70)


  # Compare all 4 weights
  for i, feature in enumerate(features):
    print(f"Weight ({feature}):   | {scratch_model.w[i]:<25.6f} | {sklearn_model.coef_[i]:<25.6f}")

  print(f"{'Bias (b)':<15} | {scratch_model.b:<25.6f} | {sklearn_model.intercept_:<25.6f}")
  print(f"{'MSE Loss':<15} | {scratch_mse:<25.6f} | {sklearn_mse:<25.6f}")
```
